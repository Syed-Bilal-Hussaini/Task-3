```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

iris = sns.load_dataset('iris')
netflix = pd.read_csv('/content/netflix_titles.csv')

iris.hist(figsize=(10, 6), color='skyblue', edgecolor='black')
plt.suptitle("Iris Numerical Distributions")
plt.show()

sns.histplot(netflix['release_year'], bins=30, kde=True, color='red')
plt.title("Netflix: Distribution of Release Years")
plt.show()
```

```python
sns.countplot(data=netflix, x='type', palette='Set2')
plt.title("Content Type on Netflix")
plt.show()

sns.countplot(data=iris, x='species')
plt.title("Iris Species Distribution")
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.boxplot(x=netflix['release_year'], color='salmon')
plt.title("Box Plot of Netflix Release Years")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(data=iris, x='species', y='sepal_width')
plt.title("Sepal Width Outliers by Species")
plt.show()
```

```python
# 1. Clean the 'duration' column to get numbers (e.g., '90 min' -> 90)
# We only do this for Movies, as TV Shows use 'Seasons'
movies_df = netflix[netflix['type'] == 'Movie'].copy()
movies_df['duration_num'] = movies_df['duration'].str.replace(' min', '').astype(float)

# 2. Create the Heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(movies_df[['release_year', 'duration_num']].corr(), annot=True, cmap='Reds')
plt.title("Correlation: Release Year vs. Movie Duration")
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.heatmap(iris.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.title("Iris Feature Correlations")
plt.show()
```
